This notebook demonstrates:
1. Exploratory analysis of synthesized e-commerce dispute data
2. Strict temporal (time-based) train/test splitting
3. Feature engineering explainability
4. Winnability classifier training (XGBoost vs Logistic Regression Baseline)
5. Cost-aware decision threshold optimization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys, os
sys.path.append('../src')

from features import build_features
from sklearn.metrics import classification_report, roc_auc_score, average_precision_score

# Load dataset
df = pd.read_csv('../data/processed/labeled_disputes.csv')
df['dispute_date'] = pd.to_datetime(df['dispute_date'])
df = df.sort_values('dispute_date').reset_index(drop=True)
print(f'Total records: {len(df)}, Win rate: {df["won_dispute"].mean():.2%}')
df.head()
```

```python
# Time-Based Train/Test Split (80% Train, 20% Held-Out Test)
split_idx = int(len(df) * 0.8)
train_df = df.iloc[:split_idx]
test_df = df.iloc[split_idx:]

X = build_features(df)
y = df['won_dispute'].values

X_train, y_train = X.iloc[:split_idx], y[:split_idx]
X_test, y_test = X.iloc[split_idx:], y[split_idx:]

print(f'Training period: {train_df["dispute_date"].min().strftime("%Y-%m-%d")} to {train_df["dispute_date"].max().strftime("%Y-%m-%d")}')
print(f'Testing period:  {test_df["dispute_date"].min().strftime("%Y-%m-%d")} to {test_df["dispute_date"].max().strftime("%Y-%m-%d")}')
```

```python
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression

# Baseline
base_model = LogisticRegression(max_iter=1000, class_weight='balanced')
base_model.fit(X_train, y_train)
base_probs = base_model.predict_proba(X_test)[:, 1]

# XGBoost Winnability Model
xgb_model = XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.05, eval_metric='aucpr')
xgb_model.fit(X_train, y_train)
xgb_probs = xgb_model.predict_proba(X_test)[:, 1]

print('Baseline PR-AUC:', average_precision_score(y_test, base_probs))
print('XGBoost  PR-AUC:', average_precision_score(y_test, xgb_probs))
print('XGBoost  ROC-AUC:', roc_auc_score(y_test, xgb_probs))
```
